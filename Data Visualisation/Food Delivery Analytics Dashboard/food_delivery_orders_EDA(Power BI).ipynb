## Part B - Controlled Cleaning

```python
df['Payment_Method'] = df['Payment_Method'].replace({
    'cash': 'Cash',
    'CARD': 'Card',
    'E-Wallet':'Wallet'
})
```

```python
df['Payment_Method'].head(20)
```

```python
conditions = [
    df['Delivery_Minutes'] > 45,
    (df['Delivery_Minutes'] > 30) & (df['Delivery_Minutes'] <= 45),
    df['Delivery_Minutes'] <= 30
]

choices = ['Delayed', 'Normal', 'Fast']

df['Delivery_Speed'] = np.select(
    conditions,
    choices,
    default='Unknown'
)
```

```python
df.head(20)
```
